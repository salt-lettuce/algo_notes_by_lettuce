## Suffix Array(接尾辞)

部分文字列を便利に扱えるデータ構造について解説します.

### Suffix Arrayとはなにか

まずは次のような配列を考えます.  
要素iに,文字列Sのi番目から末尾までの部分文字列をもつ配列.  
具体的には下記のようになります.  

文字列S = "SuffixArray" のとき
```python
suffix[0]  = "SuffixArray"
suffix[1]  = "uffixArray"
suffix[2]  = "ffixArray"
suffix[3]  = "fixArray"
suffix[4]  = "ixArray"
suffix[5]  = "xArray"
suffix[6]  = "Array"
suffix[7]  = "rray"
suffix[8]  = "ray"
suffix[9]  = "ay"
suffix[10] = "y"
```

次にこれを辞書順でソートしたときの開始位置を要素に持つ配列を`SuffixArray`といいます.  
具体的には下記のようになります.  (小文字は大文字よりも遅い)

```python
sa[0]  = 6   # "Array"
sa[1]  = 0   # "SuffixArray"
sa[2]  = 9   # "ay"
sa[3]  = 2   # "ffixArray"
sa[4]  = 3   # "fixArray"
sa[5]  = 4   # "ixArray"
sa[6]  = 8   # "ray"
sa[7]  = 7   # "rray"
sa[8]  = 1   # "uffixArray"
sa[9]  = 5   # "xArray"
sa[10] = 10  # "y"
```

SA-IS法というアルゴリズムを用いることで,文字列長をNとしたときにこの配列を`O(N)`で取得することができます.

### ライブラリでの実装

AtCoder Library(ACL)にはSuffix Arrayのライブラリが存在します。  

````python
from atcoder.string import suffix_array
S = "SuffixArray"
sa = suffix_array(S)
print(sa)
for i in sa:
    print(S[i:])
````

##

### 使い方

SuffixArrayからソートされた任意の長さの部分文字列を復元し,二分探索などに利用できます.$O(|S|N)$

````python
from atcoder.string import suffix_array
S = "SuffixArray"
sa = suffix_array(S)

S_sorted = []
N = 2
for i in sa:
    S_sorted.append(S[i:i+N])
print(S_sorted)
````
