## SortedContainers [アルゴ式](https://algo-method.com/courses/f102a001a27ba1cc)

高速に整列された入れ物を扱うことができるSortedContainersについて解説していきます。

### 

### SortedList

```python
from sortedcontainers import SortedList

sl = SortedList()
sl.add(3)
print(sl)
```

#### 操作一覧

|操作|操作内容|計算量|
|---|---|---|
|sl.add(value)|要素 `value` を追加し、ソート状態を維持する|O(logN)|
|sl.discard(value)|要素 `value` を削除する (存在しない場合もエラーにならない)|O(logN)|
|sl.remove(value)|要素 `value` を削除する (存在しない場合は `ValueError`)|O(logN)|
|sl.pop(index=-1)|`index` 番目の要素を削除して返す (デフォルトは末尾)|O(logN)|
|sl[index]|`index` 番目の要素を取得する (ランダムアクセス)|O(logN)|
|value in sl|`value` が含まれているか判定する|O(logN)|
|sl.bisect_left(value)|`value` 以上の要素が現れる最初のインデックスを返す|O(logN)|
|sl.bisect_right(value)|`value` より大きい要素が現れる最初のインデックスを返す|O(logN)|
|sl.count(value)|要素 `value` の個数を返す|O(logN)|
|sl.index(value)|要素 `value` が最初に現れるインデックスを返す|O(logN)|
|len(sl)|要素数を返す|O(1)|
|sl.clear()|全要素を削除する|O(N)|
